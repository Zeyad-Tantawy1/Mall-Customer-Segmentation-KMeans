```python
import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/Prodigy/Mall_Customers.csv')
```

```python
df.head()
```

```python
display(df.head())
display(df.info())
display(df.isnull().sum())
```

```python
X = df[['Annual Income (k$)', 'Spending Score (1-100)']]
display(X.head())
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8, 6))
plt.plot(range(1, 11), inertia, marker='o')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method to Determine Optimal K")
plt.xticks(range(1, 11))
plt.grid(True)
plt.show()
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
kmeans.fit(X)
df['Cluster Label'] = kmeans.labels_
display(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 8))
sns.scatterplot(data=df, x='Annual Income (k$)', y='Spending Score (1-100)', hue='Cluster Label', palette='viridis', s=100)
plt.title('Customer Segments based on Annual Income and Spending Score')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend(title='Cluster')
plt.grid(True)
plt.show()
```

```python
cluster_characteristics = df.groupby('Cluster Label')[['Annual Income (k$)', 'Spending Score (1-100)']].mean()
display(cluster_characteristics)
```

```python
# @title Annual Income (k$)

from matplotlib import pyplot as plt
cluster_characteristics['Annual Income (k$)'].plot(kind='line', figsize=(8, 4), title='Annual Income (k$)')
plt.gca().spines[['top', 'right']].set_visible(False)
```
